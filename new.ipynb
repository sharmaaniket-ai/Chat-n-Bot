```python
from langchain_community.document_loaders import TextLoader
Loader=TextLoader(r"C:\Users\hp\OneDrive\Documents\Desktop\new.txt")
docs=Loader.load()
```

```python
docs
```

```python
from langchain_community.document_loaders.onedrive_file import CHUNK_SIZE
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitters=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=100)
```

```python
documents=splitters.split_documents(docs)
```

```python
len(documents)
```

```python
for doc in documents:
    print(doc.page_content)
    print("--------------------------")
```

```python
from langchain_ollama import OllamaEmbeddings
embeddings=OllamaEmbeddings(model="nomic-embed-text-v2-moe", base_url="http://localhost:11434")
```

```python
embeddings.embed_documents(["Hello world", "Bye bye"])
```

```python
from langchain_community.vectorstores import FAISS
```

```python
vectorstore=FAISS.from_documents(documents, embeddings)
```

```python
vectorstore
```

```python
vectorstore.similarity_search_with_relevance_scores("what is ai?")
```

```python
retreiver=vectorstore.as_retriever()
```

```python
retreiver
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
template="""Answer the user query from the given {context}
Quesition: {question}
Answer:
"""
```

```python
prompt=PromptTemplate.from_template(template)
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
chain=({'context': retreiver, 'question': RunnablePassthrough()}) | prompt | llm | StrOutputParser()
```

```python
res=chain.invoke("What is AI and features of AI and how it works?")
```
